# 03  Figures for the paper

Every data figure of the paper and of the supplementary material is produced here from the saved
feature and result tables, so it can be regenerated without the raw data. The squared envelope
spectra with the selected peaks are produced by `01_extract_features.ipynb`
(`outputs/peak_verification/`). The test-rig photographs and the schematic figures are not produced
by the code. Each panel is saved as a separate PNG (and PDF where useful) for assembly as
subfigures in LaTeX; captions are written in LaTeX.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import Image, display

import uncertainty_core as uc

OUTPUT_DIR = Path("outputs")
cfg = uc.Config(output_dir=OUTPUT_DIR)
store = uc.FeatureStore.load(cfg.subdir("features"))
tables = cfg.subdir("tables")
figs = cfg.subdir("figures")
metrics = pd.read_csv(tables / "metrics.csv")
inference = pd.read_csv(tables / "inference.csv")
sweep = pd.read_csv(tables / "screening_sweep.csv")
ss = pd.read_csv(tables / "sample_size_check.csv")
synth = pd.read_csv(tables / "synthetic_check.csv")
draws = np.load(tables / "subsample_draws.npz")
pl = uc.Plotter(cfg)
T = cfg.representative_test
```

## 1. Spread within each assembly state (test session T2, one panel per case, shared axes)

```python
for h in uc.HEALTH_STATES:
    paths = pl.within_case(store, h, T, figs)
    for c, p in paths.items():
        print(p.name)
        display(Image(filename=str(p.with_suffix(".png")), width=420))
```

## 2. Bias across the cases (test session T2, segmented and full signal)

```python
for h in uc.HEALTH_STATES:
    paths = pl.across_case(store, h, T, figs)
    for lv, p in paths.items():
        print(p.name)
        display(Image(filename=str(p.with_suffix(".png")), width=420))
```

## 3. Heatmaps of $D_{\mathrm{hull}}$, $D_\Sigma$, $B$ and $\rho_\Sigma$

```python
hm = pl.heatmaps(metrics, figs)
for p in hm:
    if "rho_sigma" in p.name or "D_hull" in p.name:
        print(p.name)
        display(Image(filename=str(p), width=330))
print(f"{len(hm)} heatmap panels saved")
```

## 4. Intervals of the indicators

$\rho_\Sigma$ with 95 % recording-level bootstrap intervals (filled markers: permutation $p < 0.05$),
and $D_{\mathrm{hull}}$ with $\pm 2$ leave-one-recording-out jackknife standard errors.

```python
p = figs / "fig_rho_sigma_intervals_grid"
pl.rho_interval_grid(metrics, inference, p)
print(p.name); display(Image(filename=str(p.with_suffix(".png")), width=800))
for h in uc.HEALTH_STATES:
    for lv in uc.LEVELS:
        p = figs / f"fig_interval_rho_sigma_{h}_{lv}"
        pl.interval_plot(metrics, inference, "rho_sigma", h, lv, p, ref_line=1.0, ylabel=r"$\rho_\Sigma$")
        p2 = figs / f"fig_interval_D_hull_{h}_{lv}"
        pl.interval_plot(metrics, inference, "D_hull", h, lv, p2, ylabel=r"$D_{\mathrm{hull}}$")
        print(p2.name); display(Image(filename=str(p2.with_suffix(".png")), width=600))
```

## 5. Sensitivity of $D_{\mathrm{hull}}$ to the outlier-screening threshold

```python
for h in uc.HEALTH_STATES:
    p = figs / f"fig_S_screening_sweep_{h}"
    pl.screening_sweep(sweep, h, p)
    print(p.name); display(Image(filename=str(p.with_suffix(".png")), width=500))
```

## 6. Hull vs covariance dispersion, and the five-point subsampling check

```python
p = figs / "fig_dispersion_comparison"
pl.dispersion_comparison(metrics, p)
display(Image(filename=str(p.with_suffix(".png")), width=450))
for h, c in (("damaged", "C1"), ("damaged", "C3"), ("healthy", "C1")):
    row = ss[(ss.health == h) & (ss.test == T) & (ss.case == c)].iloc[0]
    p = figs / f"fig_subsample_{h}_{T}_{c}"
    pl.subsample_hist(draws[f"{h}_{T}_{c}"], row.D_hull_full, row.D_hull_seg, p)
    print(p.name); display(Image(filename=str(p.with_suffix(".png")), width=500))
```

## 7. Synthetic check and the assembly-state centroids used in the task check

```python
p = figs / "fig_S_calibration"
pl.synthetic_check(synth, p)
display(Image(filename=str(p.with_suffix(".png")), width=500))
for lv in uc.LEVELS:
    p = figs / f"fig_cluster_map_{lv}"
    pl.cluster_map(store, lv, p)
    print(p.name); display(Image(filename=str(p.with_suffix(".png")), width=450))
```
